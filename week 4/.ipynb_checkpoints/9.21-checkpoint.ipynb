## Loops 
'For' loop will iterate through every item in the list. Notice syntax

```python
for step in routine: #extract each element in routine one by one
    print(step)
```

'For' loops consist of:
- the 'for' keyword: indicates a looping structure
- an incremented variable
- the 'in' keyword followed by an iterable list/collection
- the **:** tells python that the looping lines of code are to follow. The lines after the colon will be indented 

```python
fruit = "banana"
for i in fruit: 
    print(i.upper())#upper - captilizes the characters 
```

```python
grades = [56, 59, 1, 99, 55, 80, 90]
sum_of_grades = 0 #Initializing the sum variable at zero to add to each step of loop

for grade in grades:
    sum_of_grades = sum_of_grades + grade

print(grades)
print(sum_of_grades)
print(sum_of_grades/len(grades))
```

```python
total_length = 0

for album in beatles_albums:
    running_time = album["running_time"] 
    minutes, seconds = running_time.split(":")
    running_time_in_seconds = int(minutes) * 60+ int(seconds)
    print(running_time_in_seconds)
    album["running_time_in_seconds"] = running_time_in_seconds
    total_length = total_length + album["running_time_in_seconds"]

print(total_length/len(beatles_albums))
```

## Function declarations
(Lambda: a short, unnamed function that extracts a certain keyword)

Custom functions that do specific tasks in Python


```python
def calculate_seconds(time_string): # def+function name+ arguments + colon
    #expecting string as input as represented by the placeholder "time_string"
    minutes, seconds = time_string.split(":")
    total_seconds = int(minutes)*60 + int(seconds)
    return total_seconds

print(calculate_seconds("40:45"))
```

Function declarations are made up of:
- the 'def' keyword
- function name
- arguments
- colon
- return statement 

```python
def angry(text):
    text = text.upper()
    exclamations = "!"*len(text)
    return text + exclamations

print(angry("hello world"))
```

## Conditionals

Conditionals are a fundamental structure that relies on comparison operators, which compare two values and return a value of True or False

- x == y: are the values equal
- x !=y: are the values not equal
- x < y: is x less than y
- x > y: is x greater than y
- x <= y: is x lesser than or equal to y
- x >= y: is x greater than or equal to y


Comparison operators can be combined with logical operators, which allow you to check more than one thing
- x > y and a == b: are both conditions true
- x <y or a >b: is one condition at least true

```python
x = 5
y = 6
a = 1
b = 1

print(x > y)
print(x < y)
print(x == y)
print(x < y and a==b)
print(x>y or a>b)
```

In Python, we can run a block of code conditionally. If the conditon is true, the code will run (if and else syntax)

```python
temp = 70

if temp < 60:
    print("put on a jacket")
else:
    print("no jacket needed!")
```

We can also stack conditions using the else if, or elif keyword

```python
temp = 58

if temp >= 80:
    print("wear shorts")
elif temp >= 65:
    print("wear pants")
elif temp >= 50:
    print("wear a sweater")
elif temp >= 35:
    print("wear a jacket")
else:
    print("stay home")
```

```python
temp = 70

if temp >= 65 and temp <= 75:
    print("the weather is comfortable")
elif temp < 32 or temp > 95:
    print("the weather is extreme")
else: 
    print("i dont know what the weather is")
```

```python
grade = 60

if grade > 50:
    print("the student has a D")
elif grade < 50:
    print("The student has an F")
elif grade > 60:
    print("The student has a C")
```

## Using conditionals to traverse and filter a dataset

```python
for album in beatles_albums:
    print(album["album_title"] + ":")
    for track in album["tracks"]:
        print(track)
    
```

```python
# Find sequence " love" in the song title

for album in beatles_albums:
    for track in album["tracks"]:
        if "love" in track.lower():
            print(track)
```

```python
love_songs = [] 
for album in beatles_albums:
    for track in album["tracks"]:
        if " lov" in (" "+track).lower():
            love_songs.append(track)

print(love_songs)
print(len(love_songs))
```

## CSV data

CSV is one of the most common dataset formats (a spreadsheet with rows and columns). Each row is a new line; a comma separates each column.
It usually has a header, and the subsequent rows contain the actual data.

```python
import csv
with open("moma_artists.csv", "r", encoding="utf-8") as file:# r stands for read file; encoding is helpful if there are other languages or non-Latin characters in the file
    reader = csv.reader(file)  #intializing an object that interprets CSV data
    header = next(reader) #tells reader to read the first line and move on to the next
    rows = list(reader) # takes every other row in the datatset and makies it into a list

print(header)
print(rows[0])
```

```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)   #will intrepret the csv as a list of dictionaries
    data = list(reader)   #store the CSV data as a list of dictionaries to access

print(data[-1])
```

```python
#filter by nationality

swedish_artists = []

for artist in data: 
    if artist["Nationality"] == "Swedish": 
        swedish_artists.append(artist)

print(len(swedish_artists))
print(swedish_artists)

for artist in swedish_artists[:10]:
    print(artist["DisplayName"])
```

```python
young_artists = []

for artist in data: 
    if int(artist["BeginDate"]) > 1990:
        young_artists.append(artist)

print(len(young_artists))

for artist in young_artists[:10]:
    print(artist["DisplayName"])
```
