```python
import csv
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    data = list(reader)

print(data[0])
```

```python
first_name_of_artist = name_of_artist.split(" ")[0]
print(first_name_of_artist)
```

I want to generate a list of all artists with the first name Mary

```python
artists_named_mary = []

for artist in data: 
    first_name = artist['DisplayName'].split(" ")[0]
    if first_name == "Mary":
        artists_named_mary.append(artist)

print(len(artists_named_mary))

for mary in artists_named_mary[:10]: 
    print(mary['DisplayName'])
    
```

```python
artists_with_year = []

for artist in data:
    if artist['BeginDate'] != 0:
        artists_with_year.append(artist)

count = {}

#calculating count for each year in begin date
for year in artists_with_year:
    years = year['BeginDate']
    if year in count:
        count[year] = count[year]+1
    else:
        count[year] = 1
       
#Advanced Sorting techniques
years_sorted = sorted(count, key=lambda years: count[year], reverse=True)
print(years_sorted[:10])


```

```python
#Step 1: Remove all entries with no year

artists_with_year = []

for artist in data:
    if artist['BeginDate'] != 0:
        artists_with_year.append(artist)

print(len(artists_with_year))
```

```python
year_counts = {} # empty dictionary to store year counts

for artist in artists_with_year:
    year = artist['BeginDate'] #store begin date for each artist
    if year in year_counts:
        year_counts[year] = year_counts[year]+1# using variable as a key, if true add one to existing count
    else:
        year_counts[year] = 1

sorted_years = sorted(year_counts, key= lambda year: year_counts[year])
```
