```python
import csv
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    data = list(reader)

print(data[0])
```

Simple search of the data by First Name

```python
name_of_artist = data[0]['DisplayName']
print(name_of_artist)
```

```python
first_name_of_artist = name_of_artist.split(" ")[0]
print(first_name_of_artist)
```

I want to generate a list of all artists with the first name Mary

```python
artists_named_mary = []

for artist in data: 
    first_name = artist['DisplayName'].split(" ")[0]
    if first_name == "Mary":
        artists_named_mary.append(artist)

print(len(artists_named_mary))

for mary in artists_named_mary[:10]: 
    print(mary['DisplayName'])
    
```

resorting the moma dataset by birth year ( begin year)

```python
artists_by_year = sorted(data, key=lambda artist: int(artist['BeginDate']))

print(artists_by_year[-100])
```

```python
artists_with_year= []

for artist in data:
    if artist['BeginDate'] != '0':
        artists_with_year.append(artist)

artists_by_year_filtered = sorted(artists_with_year, key=lambda artist: int(artist['BeginDate']))
print(artists_by_year_filtered[0])

```

```python
for artist in artists_by_year_filtered[:20]:
    print(artist['DisplayName'])
```
