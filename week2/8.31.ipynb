## Combining and Concatenating Strings

Strings can be combined in several ways. One is to use the + operator

```python
dog_name = "Blue"
dog_breed = "Yorkie" 
dog_age = 9

my_dog_string = "My dog's name is " + dog_name + " and he is " + str(dog_age)
print(my_dog_string)
```

```python
# F-strings 

my_dog_string = f"My dog's name is {dog_name}. He is {dog_age}, and he is a {dog_breed}"
print(my_dog_string)
```

## Cleaning up text with strip()

```python
messy_string = "        I have a bunch of unnecessary white space before and after me!   "
print(messy_string)
print(messy_string.strip())
# helpful to remove anything not needed when trying to access copied text
```

You can find the *length* of a string by using the len() function, which returns the length of characters

```python
print(len(messy_string))
clean_string = messy_string.strip()
print(len(clean_string))
```

## Parsing comma-delimited data

Python has several string methods for parsing by delimiting characters. (E.g., commas). The 'split(",") method will separate the string based on comma locations


```python
csv_data = "Nirvana, Smells like teen spirit, 5:01, 1991"

artist, song, length, year = csv_data.split(",")

my_string = f"My favorite band is {artist}, and their best song is {song.strip()}, which came out in {year} and is {length} long"
print(my_string)
```

## Slicing a string

We can extract a certain range of characters from a string using the slicing approach

```python
my_text = "saxophone"
print(my_text[2:5])
print(my_text[:3])
print(my_text[0])
print(len(my_text))
print(my_text[-1]) #prints last character by counting from the end 
```

```python
my_texts = "January 15 2005"
print(my_texts[-4:]) #refers to 4 characters from the end
```
