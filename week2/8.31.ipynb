## Combining and Concatenating Strings

Strings can be combined in several ways. One is to use the + operator

```python
dog_name = "Blue"
dog_breed = "Yorkie" 
dog_age = 9

my_dog_string = "My dog's name is " + dog_name + " and he is " + str(dog_age)
print(my_dog_string)
```

```python
# F-strings 

my_dog_string = f"My dog's name is {dog_name}. He is {dog_age}, and he is a {dog_breed}"
print(my_dog_string)
```

## Searching, counting, extracting

Some methods in Python search for substrings

```python
print("cat" in my_dog_string)
```

```python
print("dog" in my_dog_string)
```

```python
my_dog_string = my_dog_string + " I really love dogs and dogs are my best friends"
print(my_dog_string)
```

```python
dog_count = my_dog_string.count("dog")
print(dog_count)
print(f"The word 'dog' appears {dog_count} times.")
```

```python
cat_count = my_dog_string.count("cat")
print(f"The word 'cat'appears {cat_count} times")
```
