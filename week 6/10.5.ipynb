## Saving CSVs

How to save CSV from a list of dictionaries


```python
import csv

with open("moma_art.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    data = list(reader)

print(data[0].keys())
```

```python
#filtering the dataset into smaller dictionaries

yoko_ono_artworks = []

for artwork in data:
    if artwork['Artist'] == "Yoko Ono":
        yoko_ono_artworks.append(artwork)

print(len(yoko_ono_artworks))
```

```python
with open("yoko_ono_artworks.cs", "w", encoding ="utf-8") as file: #w for writing a new csv
    writer = csv.DictWriter(file, yoko_ono_artworks[0].keys())#specifying keys in yoko_ono_artworks as the columns in the CSV
    writer.writeheader() #create the Headers row in the CSV
    writer.writerows(yoko_ono_artworks) # going through each row in the csv
```

## Regular Expressions

Regular Expressions are a markup language used for identifying patterns in strings, in order to do complex searches. They use a bunch of specific special characters

```python
caption = "William Wegman, Cotto, 1970. Gelatin silver print, 7 7/8 × 7 3/4 in. (20 × 19.7 cm). Whitney Museum of American Art, New York; purchase with funds from the Mrs. Percy Uris Purchase Fund and the Photography Committee 92.14 © William Wegman"
```

```python
import re

match = re.search(r"([0-9]{4})",  caption)

print(match)

year = match.group(1)

print(year)
```

```python
import csv

with open("../week5/moma_artists.csv", "r", encoding = "utf-8") as file:
    reader = csv.DictReader(file)
    artists_data = list(reader)

print(artists_data[0])
    
```

```python
catherines = []

for artist in artists_data: #^ means beginning of the string , w* anythign else afterwards
    if re.search(r"^[CK]ath\w*", artist['DisplayName']):
        catherines.append(artist)

print(len(catherines))
for catherine in catherines:
    print(catherine['DisplayName'])
```

## API

APIs are used to query open data providers in real time by grabbing data from a server. The query is provided through a URL, and the response is usually formatted as a JSON object. The APIs we will be looking at:
- The Met API: returns object IDs for search items and returns details of a given object based on an object ID
- NYTTimes API: can return lists of metadata on articles based on search terms
- 

```python
import requests #brings in functions needed to grab the data from the web and deal with responses 

#variables that make up the components of the URL to query MET API
base_url = "https://collectionapi.metmuseum.org" #same no matter what
search_path = "/public/collection/v1.1/search" #this is the "endpoint" that deals with searching object IDS in the question 
search_query = "?q=Antelope" #parametters of the API query, q means: search term, 
date_range = "&dateBegin=1900&dateEnd=2000"

url = f'{base_url}{search_path}{search_query}{date_range}'
search_response = requests.get(url)

print(url)  
```

```python
print(type(search_response))
```

```python
print(dir(search_response))
```

```python
print(search_response.ok)
```

```python
print(search_response.status_code) #404 means not found, 200 means okay
```

```python
met_objects = search_response.json()

print(type(met_objects))
```

```python
print(met_objects.keys()) #shows keys contained in the JSON data
```

```python
print(met_objects['total']) #total key (total amount of objects with the search criteria
```

```python

antelope_objects_ids = met_objects['objectIDs']
print(len(antelope_objects_ids)) #this contains a list of MET Object IDs representing Antelope-related stuff
```

```python
object_path = "/public/collection/v1/objects/"
antelope_item = antelope_objects_ids[0]
url = f"{base_url}{object_path}{antelope_item}"
print(url)
response_object = requests.get(url)

print(response_object.ok)
```

```python
antelope_object_details = response_object.json()

print(antelope_object_details.keys())
```

## Below is an experiment to pull ALL object details about ALL antelope objects in my objectID list


```python
from time import sleep 

all_antelope_details = []


for antelope_id in antelope_object_ids:
    url = f"{base_url}{object_path}{antelope_id}"
    response_object = requests.get(url)
    all_antelope_details.append(response_object.json())
    sleep(0.5)

    
```
