```python
import csv

with open("moma_art.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    data = list(reader)

print(data[0].keys())


pablo_picasso = []

for item in data:
    if item['Artist'] == "Pablo Picasso":
         pablo_picasso.append(item)  


print(len(pablo_picasso))
```

```python
with open("pablo_picasso.cs", "w", encoding ="utf-8") as file: 
    writer = csv.DictWriter(file, pablo_picasso[0].keys())
    writer.writeheader() 
    writer.writerows(pablo_picasso) 
```
