## Filtering data

We can filter dataframes using comparison operators like ==
Comparing a column to a value will return a series of every row in the column, and whether that matches the criteria


```python
df['title'] == 'Woman'
```

```python
woman_mask = df['title'] == 'Woman'

type(woman_mask)
```

```python
df[woman_mask].info()

```

The above series is called a *mask*, which identifies the filtered content as rows flagged as True or False. We can use that *mask* to query the DataFrame, which will return a DataFrame that has been filtered according to the mask.

We can also use `df['title'].str.contains('Woman', na = False)`

We need to include na=False to handle missing values, otherwise the mask would inherit those incorrectly

```python
courbet = df['artistDisplayName'].str.contains('Courbet', na=False)

df[courbet]
```

## Graphing or Plotting data

Pandas can generate graphs and charts using `.plot()`. We can generate a variety of types of graphs, including: 

-'bar'
-'barh' (horizontal bar)
- 'pie'
- 'line'
- 'hist'
- 'scatter'

Bar charts are useful when dealing with the value_counts for a given series

```python
df.value_counts('department').plot(kind='barh')
```

```python
df.value_counts('country').plot(kind = 'pie')
```

```python
df.value_counts('artistDisplayName').nlargest(10).plot(kind = 'barh', ylabel='artist name', title='works with women')
```
